```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
vdf = pd.read_csv("beeActivity.csv")
```

```python
vdf = vdf[vdf['group_id'] == 11]
vdf
```

```python
x = vdf['cx']
y = vdf['cy']
im = plt.imread("final.jpg")
implot = plt.imshow(im)
plt.plot(x, y, c="red")
plt.annotate(
    "",
    xy=(x.iloc[-1], y.iloc[-1]),        
    xytext=(x.iloc[-2], y.iloc[-2]),   
    arrowprops=dict(
        arrowstyle="->",
        color="red",
        linewidth=2,
    ),
)
plt.ylim(1450, 0) 
plt.gca().invert_yaxis()
plt.gca().invert_xaxis()
plt.savefig(
    "trajectory.png",
    dpi=300,
    bbox_inches="tight",
    pad_inches=0,
)

```

```python
x = vdf['frame']
y = vdf['cy']
plt.plot(x, y, c="red")
plt.ylim(1450, 0) 
plt.gca().invert_yaxis()
plt.gca().invert_xaxis()
plt.xlabel("Frame")
plt.ylabel("cy")
plt.savefig(
    "trajectoryYlim.png",
    dpi=300,
    bbox_inches="tight",
    pad_inches=0,
)
```
